```python
import csv
with open("emp.csv","w",newline='') as f:
  w=csv.writer(f)
  w.writerow(['ENO','ENAME','ESALARY','EADDR'])
  n=int(input("enter the nuimmber of employees"))
  for i in range (n):
    eno=input("enter employee no :")
    ename=input("enter employee name :")
    esal=input("enter employee salary :")
    eaddr=input("enter employee address :")
    w.writerow([eno,ename,esal,eaddr])
print("data inserted succeessfully")
```

```python
import numpy as np
students=np.array([
    [85,90,70],
    [48,78,49],
    [95,50,40],
    [25,40,10]

])
students
```

```python
avg_m=np.mean(students,axis=1)
print("average Marks : " , avg_m)
```

```python
avg_m=np.median(students,axis=0)
print("average Marks : " , avg_m)
```

```python
avg_m=np.median(students,axis=1)
print("average Marks : " , avg_m)
```

```python
topper=np.argmax(students,axis=1)
print("Toppper Student : " ,topper)
```

```python
failed_student=np.where(avg_m<50)
print("failed Student : " ,failed_student[0]+1)
```

```python
for i, avg in enumerate(avg_m):
  status="pass" if avg > 50 else "Fail"
  print(f"student {i+1}: avg ={avg:.2f} | {status}")
```
