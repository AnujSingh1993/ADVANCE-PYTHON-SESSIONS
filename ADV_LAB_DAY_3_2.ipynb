```python
import matplotlib.pyplot as plt

# Bar Chart
months = ['Jan', 'Feb', 'Mar', 'April']
sales = [250, 300, 400, 350]

plt.bar(months, sales, color='skyblue')
plt.title('Monthly sales')
plt.xlabel('slaves')
plt.ylabel('masters')
plt.show()
```

```python
plt.barh(months, sales, color='orange')
plt.title('Monthly sales (Horizontal)')
plt.xlabel('sales')
plt.ylabel('months')
plt.show()
```

```python
import matplotlib.pyplot as plt

# Scatter Plot (comparing Height vs Weight)
height = [150, 160, 170, 180,175]
weight = [50, 60, 65, 75,87]

plt.scatter(height, weight, color='green')
plt.title('Height vs Weight')
plt.xlabel('Height (cm)')
plt.ylabel('Weight (kg)')
plt.show()
```

```python
import matplotlib.pyplot as plt

# Pie Chart (Market Share)
brands = ['Brand A', 'Brand B', 'Brand C']
shares = [45, 30, 25]

plt.pie(shares, labels=brands, autopct='%1.1f%%', startangle=90)
plt.title('Market Shares')
plt.axis('equal')
plt.show()
```

```python
import matplotlib.pyplot as plt

# Multiple Plot in one figure
x = [1, 2, 3, 4]
y1 = [1, 4, 9, 16]
y2 = [1, 2, 3, 4]

plt.plot(x, y1, label='Square')
plt.plot(x, y2, label='Linear')

plt.legend()
plt.title('Multiple Lines')
plt.show()
```

```python
import matplotlib.pyplot as plt

# Subplots
fig, axs = plt.subplots(2, 2)

axs[0, 0].plot(x, y1)
axs[0, 0].set_title('Square')

axs[0, 1].bar(months, sales)
axs[0, 1].set_title('Sales')

axs[1, 0].scatter(height, weight)
axs[1, 0].set_title('Scatter')

axs[1, 1].pie(shares, labels=brands, autopct='%1.1f%%')
axs[1, 1].set_title('Pie Chart')

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

data = {
    'name': ['Anuj','Boycarti','Cudie','Drake','El nino'],
    'math': [90,67,90,56,88],
    'science' : [85,99,90,46,68],
    'english' : [92,47,80,96,18]
}
df=pd.DataFrame(data)
print(df)
```

```python
marks=df[['math','science','english']].values
average_marks=np.mean(marks,axis=1)
df['marks']=average_marks
print(df)
```

```python
def grade(avg):
  if avg >= 90:
    return "a"
  elif avg >= 75:
    return "B"
  elif avg >= 60:
    return "C"
  else:
    return "D"
```

```python
df['Grade']=df['marks'].apply(grade)
df
```

```python
import matplotlib.pyplot as plt

plt.figure()
plt.bar(df["name"], df["marks"])
plt.xlabel('Students')
plt.ylabel('Average Marks')
plt.title('Average Marks of Students')
plt.show()
```

```python
# Average marks per subject

df[["math", "science", "english"]].mean().plot(kind="bar")

plt.title("Average Marks per Subject")
plt.ylabel("Marks")
plt.show()
```
